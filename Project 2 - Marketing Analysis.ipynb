```python
# Create Age group buckets as that column seems to be important for segmentation
```

```python
# sorting to see all ages and overall range
food['Age'].sort_values().head(n=50)
```

```python
# define age groups
age_groups = [(23,30),(31,40),(41,50),(51,60),(61,70),(71,85)]
# we will assign people from the dataset based on their age, on the above list
def assign_age_group(Age):
    for age_range in age_groups:
        if age_range[0] <= Age <= age_range[1]:
           return f"{age_range[0]}-{age_range[1]}"
    return ('Unknown')
```

```python
food['Age_group'] = food['Age'].apply(assign_age_group)
```

```python
# large population in mid age groups and low in the two end groups
food['Age_group'].value_counts()
```

```python
food.head()
```
